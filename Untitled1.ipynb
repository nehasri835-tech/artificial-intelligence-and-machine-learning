```python

from sklearn.linear_model import LinearRegression
import numpy as np

# Sample data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

# Train the model
model = LinearRegression()
model.fit(X, y)

# Predict
prediction = model.predict([[6]])

print("Prediction for 6:", prediction[0])
```
